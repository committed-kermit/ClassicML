```python
from sklearn.datasets import make_regression

import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt
import plotly.express as px
import plotly.graph_objects as go

from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
X,y=make_regression(n_samples=100,n_features=2,n_informative=2,n_targets=1,noise=50)
```

```python
df=pd.DataFrame({'feature 1':X[:,0],'feature 2':X[:,1],'target':y})
```

```python
df.head()
```

```python
df.sample(15)
```

```python
fig=px.scatter_3d(df,x='feature 1',y='feature 2',z='target')
fig.show()
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=3)
```

```python
lr=LinearRegression()
```

```python
lr.fit(X_train,y_train)
```

```python
y_pred=lr.predict(X_test)
```

```python
print('MAE:',mean_absolute_error(y_test,y_pred))
print('MSE:',mean_squared_error(y_test,y_pred))
print('RMSE:',np.sqrt(mean_squared_error(y_test,y_pred)))
```

```python
x = np.linspace(-5, 5, 10)
y = np.linspace(-5, 5, 10)
xGrid, yGrid = np.meshgrid(y, x)

final = np.vstack((xGrid.ravel().reshape(1,100),yGrid.ravel().reshape(1,100))).T
z_final = lr.predict(final).reshape(10,10)

z = z_final
```

```python
fig = px.scatter_3d(df, x='feature 1', y='feature 2', z='target')

fig.add_trace(go.Surface(x = x, y = y, z =z, opacity=0.3 ))

fig.show()
```

```python
y_pred_original_data = lr.predict(X)

fig.add_trace(go.Scatter3d(x=df['feature 1'], y=df['feature 2'], z=y_pred_original_data, mode='markers', marker=dict(color='red', size=3), name='Predicted Points on Plane'))

fig.show()
```
